## Project Objective
The goal of this project is to develop an autonomous food delivery robot for restaurants. The robot is designed to deliver food accurately to customers’ tables. Each table has a unique QR code, and when an order is ready, the robot identifies and follows the corresponding QR code to deliver the food to the correct table.


## Contributions
There are currently limited implementations of autonomous food delivery robots using QR-code-based navigation. By developing this system, we will demonstrate a simple and reliable approach for accurate table-level delivery.
This project will also allow us to test algorithms in simulation and deploy them on a real robot, since the simulated model will closely match the physical system.


## Project Plan
The textbooks or websites include information on building a custom robot. We will explore the ROS website to learn how to model the cameras and autonomous vehicles in the class, Moreover, we will use the class’s robot specifications to identify the robot’s dimensions, weight, and robot performance.
